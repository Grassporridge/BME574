## A. The experiment in plain English

The same **57 strains** appear in every simulation. Their intrinsic growth rates, underlying susceptibility values and equal starting abundances stay fixed. We vary the number of resistant strains: **0, 1, 5, 10, 20, 30, 40, 57**.

For each intermediate count, there are **10 resistance-assignment orders**. An order is a shuffled list of strain identities. The first k strains become carriers when the count is k; raising the count adds carriers without removing earlier ones. The orders change *who* is resistant, not the underlying strain traits. The zero- and all-resistant cases each have one possible assignment and are included once.

Each configuration has antibiotic-treated and untreated versions. The dynamics are deterministic: repeating the exact same configuration gives the same trajectory. Between-order variation is **carrier-identity variation**, not technical replicate noise.

| Time | What happens |
|---|---|
| 0 h | Inoculate at total 0.005 OD, divided equally among strains |
| 0–4 h | Grow without antibiotics |
| **4 h** | **Pretreatment baseline**, immediately before adding drug |
| 6 h | Early response, two hours after drug addition |
| 24 h | End of exposure; the sampled community is **before dilution** |
| Immediately after 24 h | 1:100 transfer into fresh drug-free medium |
| 48 h | Recovery endpoint |

All strains consume the same nutrient. Carriers pay a constant growth cost, are immune to antibiotic killing and degrade drug in the shared medium. Thus intrinsic traits stay fixed, but effective maximum growth changes when a strain receives resistance.

**Two important distinctions**

- Increasing carrier count also increases initial resistant biomass. This design does not isolate distribution while holding resistant biomass constant.
- The model does not track enzyme concentration. Resistant-carrier biomass is a **proxy for β-lactamase capacity**, not measured gene or protein abundance.

**Sampling precision:** the uploaded notebook uses the original stored trajectories. When an exact requested time is absent, it uses the nearest saved point. In particular, the original 48 h record can actually be 47.9 h. This reorganization preserves that behavior rather than silently changing the simulations. `samples.actual_hour` records the actual value and the timing audit below reports it.


## A1. Biological rules, parameters and assumptions

There are 57 strains competing for one shared nutrient. Let Nᵢ be strain biomass (OD), R nutrient (OD-equivalents), a antibiotic concentration (µg/mL), and Iᵢ = 1 for a resistant carrier and 0 otherwise.

**Growth cost and resource-dependent growth**

$$\mu_i=\mu_i^{\mathrm{intrinsic}}(1-cI_i),\qquad g_i=\mu_i\frac{R}{K+R}$$

The intrinsic rate is the maximum growth rate before paying for resistance; μᵢ is the maximum after that cost. The realized growth rate gᵢ also depends on food. The default carrier cost c = 0.05 means a 5% reduction. The resource factor R/(K + R) goes from 0 toward 1. At R = K = 0.05, growth is half its food-unlimited rate. A faster intrinsic grower still slows as nutrient runs out.

**Growth-dependent antibiotic killing and biomass change**

$$k_i=(1-I_i)\gamma\frac{(a/IC50_i)^h}{1+(a/IC50_i)^h}g_i,\qquad \frac{dN_i}{dt}=(g_i-\delta-k_i)N_i$$

Here kᵢ is the antibiotic killing rate per hour. IC50ᵢ is the concentration giving half the maximum drug-dependent killing factor at fixed growth; h controls how sharply that factor rises. Defaults: γ = 2.5 scales killing relative to growth, Hill exponent h = 2, and background death δ = 0.004/h. Growth adds biomass; both death terms subtract it. Carriers are immune to antibiotic killing in this model, but still experience background death. Nutrient-starved cells grow slowly and therefore experience little growth-dependent killing.

**Nutrient consumption and shared antibiotic removal**

$$\frac{dR}{dt}=-\sum_iN_i g_i,\qquad \frac{da}{dt}=-v_{\max} B_R\frac{a}{K_m+a}$$

Here B_R is live resistant biomass. Defaults: v_max = 10,000 µg/mL/h per OD and K_m = 5.3 µg/mL. All carriers have the same per-OD degradation capacity. The drug-removal fraction a/(K_m + a) is half its maximum at a = K_m; it approaches 1 at high drug concentration. Growth consumes nutrient with yield one in these units; death does not recycle it. There is no explicit enzyme pool, spontaneous drug decay, cross-feeding, spatial structure, mutation or gene transfer.


## A2. Fixed traits, transfer and numerical methods

**How the fixed strain traits were drawn**

For growth, σ = √ln(1 + CV²), CV = 0.08, and μᵢᶦⁿᵗʳⁱⁿˢⁱᶜ = 0.55 exp(σZᵢ − σ²/2) /h. For susceptibility, IC50ᵢ = 5 exp(0.70Z′ᵢ − 0.70²/2) µg/mL. Z and Z′ are standard-normal draws from the fixed seed 42. This produces positive lognormal traits.

**Code-versus-comment detail:** the actual IC50 formula makes 5 µg/mL its distribution mean, although an original source comment calls it the median. The formula is preserved here; the code has not been changed to match the comment. Sample averages need not equal the distribution means.

**Protocol and numerics:** total inoculum 0.005 OD; initial nutrient 1.0 OD-equivalent; antibiotic dose 50 µg/mL at 4 h; 1:100 proportional transfer at 24 h into fresh nutrient 1.0 with zero drug; recovery ends nominally at 48 h. The ODE uses LSODA and log biomass, relative tolerance 10⁻⁸, absolute tolerance 10⁻¹⁰, and a nominal 0.1 h output spacing. A strain crossing the one-cell floor (10⁻⁹ OD at 10⁹ cells/OD) is removed irreversibly. Transfer removes strains falling below that floor. These deterministic rules do not add Gillespie randomness.

**The dilution step, explicitly.** If minus means immediately before transfer and plus means immediately afterward:

$$N_i(24^+)=0.01N_i(24^-),\qquad R(24^+)=1,\qquad a(24^+)=0$$

Any transferred strain below the model's one-cell floor is set to zero. Without that removal, proportional dilution preserves relative composition and diversity even though total biomass drops 100-fold. Fresh nutrient allows survivors to regrow; the drug-free recovery phase is a specified experimental reset, not spontaneous disappearance of antibiotic.

These are specified model assumptions and input values, not parameters inferred by fitting the recovery plots.


## B. Metric guide: what the numbers mean

Let $N_i(t)$ be strain i's absolute biomass in OD, $B(t)=\sum _iN_i(t)$ the total, and $p_i(t)=N_i(t)/B(t)$ its relative abundance.

| Metric | Calculation | Meaning and scale |
|---|---|---|
| Total biomass | $B=\sum _iN_i$ | Actual community size in OD; not normalized |
| Resistant biomass | $B_R=\sum _{i:\mathrm{carrier}}N_i$ | Actual biomass belonging to resistant strains |
| Resistant percentage | $100B_R/B$ | 0–100%; who makes up the community, not its total size |
| Shannon diversity | $H=-\sum _{p_i>0}p_i\ln  p_i$ | 0 for one strain; maximum $\ln (57)\approx 4.04$ for equal abundances |
| Gini–Simpson diversity | $S=1-\sum _i p_i^2$ | Probability that two independent cell selections belong to different strains; 0 to $1-1/57\approx 0.982$ here |
| Shannon retention | $H(t)/H(4)$ | 1 unchanged; below 1 lower; above 1 higher. Not the fraction of strains surviving |
| Simpson retention | $S(t)/S(4)$ | Same interpretation, using Gini–Simpson |
| Bray–Curtis to baseline | $BC(t,4)=\frac{1}{2}\sum _i\lvert p_i(t)-p_i(4)\rvert $ | 0 identical composition; 1 no overlap. Higher = more restructuring |
| Biomass versus control | $B_T(t)/B_C(t)$ | 1 means equal biomass to the matched untreated community |

T means treated; C means untreated. Relative compositions are normalized before calculating Bray–Curtis, so it measures composition independently of total biomass.

**Do not equate these metrics.** A community can lose most of its biomass while keeping similar proportions and diversity. Conversely, biomass can recover while different strains dominate. Low Bray–Curtis is not evidence of high biomass.

**Empty communities:** if total biomass is zero, relative composition and diversity are undefined. They are not entered as zero diversity or zero dissimilarity. Retention is also undefined if its baseline index is zero. Detected strain counts use the original model's relative threshold; non-detection is not automatically extinction.

```python
SIM_KEYS = ['t','N','R','a','carrier','mu','ic50','pre_transfer','post_transfer','budget_error','t_transfer']
INITIALIZER_VERSION = 'fixed_traits_explicit_mask_v1'

def cached_simulation(mask,treated):
    p=replace(BASE,n_resistant=int(mask.sum()),dose=DOSE if treated else 0.)
    payload=dict(model=MODEL_SHA256,initializer=INITIALIZER_VERSION,
                 parameters=vars(p),carrier_mask=mask.tolist())
    key=hashlib.sha256(json.dumps(payload,sort_keys=True).encode()).hexdigest()
    path=CACHE/(key+'.npz')
    if path.exists() and not RECOMPUTE:
        with np.load(path,allow_pickle=False) as saved:
            sim={key:saved[key].copy() for key in SIM_KEYS}
        sim['params']=p
    else:
        with assigned_carriers(mask): sim=model.simulate(p)
        temporary=path.with_suffix('.tmp')
        with temporary.open('wb') as stream:
            np.savez_compressed(stream,**{key:sim[key] for key in SIM_KEYS})
        temporary.replace(path)
    # Audit actual outputs, including cached results.
    np.testing.assert_array_equal(sim['carrier'],mask)
    np.testing.assert_allclose(sim['ic50'], IC50, rtol=1e-13, atol=0)
    np.testing.assert_allclose(sim['mu'],INTRINSIC*(1-p.resistance_cost*mask),rtol=1e-14)
    np.testing.assert_allclose(sim['N'][0],N0,rtol=1e-10)
    assert np.isfinite(sim['N']).all() and sim['N'].min()>=0
    assert abs(float(sim['budget_error']))<1e-6
    post=sim['pre_transfer']/p.dilution;post[post<p.floor]=0
    np.testing.assert_allclose(sim['post_transfer'],post,atol=1e-15)
    return sim

```

```python
RUNS, audit_rows = {}, []
for d in design.itertuples():
    for treated in [False, True]:
        sim = cached_simulation(MASKS[d.scenario], treated)
        RUNS[(d.scenario, treated)] = sim
        audit_rows.append(dict(scenario=d.scenario, treated=treated,
            n_resistant=d.n_resistant, initial_biomass=sim['N'][0].sum(),
            budget_error=float(sim['budget_error'])))
    np.testing.assert_allclose(
        model.snapshot(RUNS[(d.scenario, True)], 4),
        model.snapshot(RUNS[(d.scenario, False)], 4), rtol=1e-12, atol=1e-15)
assert model.community is original_community
audit = pd.DataFrame(audit_rows)
audit.to_csv(OUT/'simulation_audit.csv', index=False)
print(f'{len(RUNS)} simulated or cached trajectories ready; fixed initialization and matched baselines verified.')
```

## D. Calculate everything once

The next two collapsed cells define reusable calculations and plotting functions. They build four shared tables:

- `curves`: biomass, Shannon, Simpson and Bray–Curtis at every stored time.
- `samples`: those metrics at the requested sampling times, plus actual stored times and resistant/susceptible biomass.
- `paired`: treated outcomes compared with their own baselines and matched untreated controls.
- `strains`: each strain's contribution, growth traits and carrier status in each valid endpoint configuration.

All figures use these tables. This replaces repeated imports and slightly different versions of the same calculation. Individual visualizations are produced by short calls below. Full functions remain available to inspect or edit.


```python
# One set of calculations used by every figure.
def note(text):
    display(Markdown('**What these results show:** '+text))

def show(fig,name):
    fig.savefig(FIGURES/(name+'.png'),dpi=180,bbox_inches='tight')
    export_report_figure(fig,name)
    plt.show();plt.close(fig)

def snapshot_index(sim,hour):
    hits=np.flatnonzero(np.isclose(sim['t'],hour,atol=1e-7))
    return int(hits[0]) if len(hits) else int(np.argmin(abs(sim['t']-hour)))

def snapshot(sim,hour):
    return sim['N'][snapshot_index(sim,hour)]

def ratio(a,b):
    return float(a/b) if np.isfinite(a) and np.isfinite(b) and b>0 else np.nan

def bray(x,y):
    if x.sum()<=0 or y.sum()<=0:return np.nan
    return float(.5*np.abs(x/x.sum()-y/y.sum()).sum())

def profile(x,carrier):
    total=float(x.sum())
    if total<=0:
        return dict(biomass=0.,shannon=np.nan,simpson=np.nan,resistant_share=np.nan,
                    resistant_biomass=0.,susceptible_biomass=0.,detected=0,susceptible_detected=0,extinct=True)
    p=x/total;positive=p>0;detected=(p>=BASE.detection_rel)&positive
    return dict(biomass=total,shannon=float(-(p[positive]*np.log(p[positive])).sum()),
        simpson=float(1-np.sum(p*p)),resistant_share=float(x[carrier].sum()/total),
        resistant_biomass=float(x[carrier].sum()),susceptible_biomass=float(x[~carrier].sum()),
        detected=int(detected.sum()),susceptible_detected=int((detected&~carrier).sum()),extinct=False)

def build_tables():
    curves=[];sample_rows=[];paired_rows=[];strain_rows=[]
    for d in design.itertuples():
        metadata=dict(scenario=d.scenario,order_id=d.order_id,n_resistant=d.n_resistant,
                      reference=d.reference,mask_id=d.mask_id)
        baseline_by_arm={}
        for treated in [False,True]:
            sim=RUNS[(d.scenario,treated)];carrier=np.asarray(sim['carrier'],bool)
            baseline=snapshot(sim,BASELINE_HOUR);baseline_by_arm[treated]=baseline
            baseline_metrics=profile(baseline,carrier)
            x=sim['N'];total=x.sum(axis=1)
            p=np.divide(x,total[:,None],out=np.zeros_like(x),where=total[:,None]>0)
            logp=np.zeros_like(p);np.log(p,out=logp,where=p>0)
            H=-(p*logp).sum(axis=1);S=1-(p*p).sum(axis=1)
            BC=.5*np.abs(p-baseline/baseline.sum()).sum(axis=1) if baseline.sum()>0 else np.full(len(total),np.nan)
            H[total<=0]=np.nan;S[total<=0]=np.nan;BC[total<=0]=np.nan
            curves.append(pd.DataFrame(dict(**metadata,treated=treated,hour=sim['t'],record_index=np.arange(len(total)),
                biomass=total,shannon=H,simpson=S,bray_curtis=BC)))
            for hour in sorted(set([0.,BASELINE_HOUR,ENDPOINT_HOUR]+SAMPLE_HOURS)):
                idx=snapshot_index(sim,hour);m=profile(x[idx],carrier)
                sample_rows.append(dict(**metadata,treated=treated,hour=hour,actual_hour=float(sim['t'][idx]),**m,
                    shannon_retention=ratio(m['shannon'],baseline_metrics['shannon']),
                    simpson_retention=ratio(m['simpson'],baseline_metrics['simpson']),bray_curtis=bray(x[idx],baseline),
                    baseline_carrier_biomass_OD=float(baseline[carrier].sum()),
                    baseline_carrier_share=baseline_metrics['resistant_share']))
        t=RUNS[(d.scenario,True)];c=RUNS[(d.scenario,False)];mask=t['carrier']
        bT=baseline_by_arm[True];bC=baseline_by_arm[False]
        mT0=profile(bT,mask);mC0=profile(bC,mask)
        for hour in SAMPLE_HOURS:
            xt=snapshot(t,hour);xc=snapshot(c,hour);mt=profile(xt,mask);mc=profile(xc,mask)
            ht=ratio(mt['shannon'],mT0['shannon']);hc=ratio(mc['shannon'],mC0['shannon'])
            bt=bray(xt,bT);bc=bray(xc,bC)
            paired_rows.append(dict(**metadata,hour=hour,baseline_carrier_biomass_OD=float(bT[mask].sum()),
                baseline_carrier_share=mT0['resistant_share'],shannon_retention=ht,
                simpson_retention=ratio(mt['simpson'],mT0['simpson']),bray_curtis_to_baseline=bt,
                adjusted_shannon_retention=ratio(ht,hc),excess_bray_curtis=bt-bc,
                control_shannon_retention=hc,control_bray_curtis_to_baseline=bc,
                biomass_ratio_to_control=ratio(mt['biomass'],mc['biomass']),
                resistant_share=mt['resistant_share'],treated_extinct=mt['extinct'],
                treated_biomass=mt['biomass'],susceptible_detected_fraction=ratio(mt['susceptible_detected'],57-d.n_resistant)))
        endT=snapshot(t,ENDPOINT_HOUR);endC=snapshot(c,ENDPOINT_HOUR)
        if min(bT.sum(),bC.sum(),endT.sum(),endC.sum())>0:
            deltaT=endT/endT.sum()-bT/bT.sum();deltaC=endC/endC.sum()-bC/bC.sum()
            strain_rows.append(pd.DataFrame(dict(**metadata,strain=NAMES,carrier=mask,
                intrinsic_growth=INTRINSIC,effective_growth=t['mu'],ic50=t['ic50'],
                baseline_relative=bT/bT.sum(),endpoint_relative=endT/endT.sum(),
                baseline_absolute=bT,endpoint_absolute=endT,untreated_endpoint_absolute=endC,
                relative_shift=deltaT,BC_contribution=.5*np.abs(deltaT),
                adjusted_relative_shift=deltaT-deltaC,excess_BC_contribution=.5*(np.abs(deltaT)-np.abs(deltaC)))))
    return pd.concat(curves,ignore_index=True),pd.DataFrame(sample_rows),pd.DataFrame(paired_rows),pd.concat(strain_rows,ignore_index=True)

def make_summaries():
    global ranking,growth_reference,profiles,status_comparison,correlations,by_order
    intermediate=strains[~strains.reference]
    ranking=intermediate.groupby('strain').agg(mean_BC_contribution=('BC_contribution','mean'),
        mean_relative_shift=('relative_shift','mean'),mean_adjusted_shift=('adjusted_relative_shift','mean'),
        mean_excess_contribution=('excess_BC_contribution','mean'),intrinsic_growth=('intrinsic_growth','first'),
        ic50=('ic50','first'),carrier_frequency=('carrier','mean')).sort_values('mean_BC_contribution',ascending=False)
    growth_reference=pd.DataFrame(dict(strain=NAMES,intrinsic_growth=INTRINSIC)).set_index('strain')
    growth_reference['growth_rank']=growth_reference.intrinsic_growth.rank(ascending=False,method='min').astype(int)
    growth_reference['growth_vs_average_percent']=100*(growth_reference.intrinsic_growth/INTRINSIC.mean()-1)
    profiles=intermediate.groupby(['strain','carrier']).agg(n_configurations=('scenario','size'),
        effective_growth=('effective_growth','mean'),mean_BC_contribution=('BC_contribution','mean'),
        mean_baseline_share=('baseline_relative','mean'),mean_endpoint_share=('endpoint_relative','mean'),
        mean_relative_shift=('relative_shift','mean')).reset_index().merge(growth_reference.reset_index(),on='strain')
    profiles['status']=np.where(profiles.carrier,'Resistant','Susceptible')
    profiles['change_percentage_points']=100*profiles.mean_relative_shift
    z=intermediate[intermediate.n_resistant.eq(STATUS_COUNT)].groupby(['strain','carrier']).agg(
        mean_adjusted_shift=('adjusted_relative_shift','mean'),n=('scenario','size')).unstack('carrier')
    z.columns=[f'{a}_{"R" if b else "S"}' for a,b in z.columns]
    if {'n_R','n_S','mean_adjusted_shift_R','mean_adjusted_shift_S'}.issubset(z.columns):
        status_comparison=z[(z.n_R>=MIN_PER_STATUS)&(z.n_S>=MIN_PER_STATUS)].copy()
        status_comparison['R_minus_S']=status_comparison.mean_adjusted_shift_R-status_comparison.mean_adjusted_shift_S
    else:status_comparison=pd.DataFrame()
    groups=[];within=[]
    draws=np.random.default_rng(BOOTSTRAP_SEED).integers(0,N_ORDERS,size=(BOOTSTRAPS,N_ORDERS))
    for metric in ['shannon_retention','bray_curtis_to_baseline','adjusted_shannon_retention','excess_bray_curtis']:
        for hour in SAMPLE_HOURS:
            g=paired[(~paired.reference)&paired.hour.eq(hour)].dropna(subset=[metric,'baseline_carrier_biomass_OD'])
            g=g[g.baseline_carrier_biomass_OD>0];blocks=[]
            for order in range(N_ORDERS):
                q=g[g.order_id.eq(order)];blocks.append(q[['baseline_carrier_biomass_OD',metric]].to_numpy())
                within.append(dict(metric=metric,hour=hour,order_id=order,rho=safe_rho(q.baseline_carrier_biomass_OD,q[metric])))
            boot=[]
            for ids in draws:
                a=np.concatenate([blocks[i] for i in ids]);boot.append(safe_rho(a[:,0],a[:,1]))
            boot=np.array(boot);boot=boot[np.isfinite(boot)]
            lo,hi=np.quantile(boot,[.025,.975]) if len(boot) else (np.nan,np.nan)
            groups.append(dict(metric=metric,hour=hour,n=len(g),distinct_masks=g.mask_id.nunique(),
                orders=g.order_id.nunique(),rho=safe_rho(g.baseline_carrier_biomass_OD,g[metric]),low=lo,high=hi,
                pearson_r_log10_x=float(pearsonr(np.log10(g.baseline_carrier_biomass_OD),g[metric]).statistic)
                if len(g)>2 and np.ptp(g[metric])>1e-12 else np.nan))
    correlations=pd.DataFrame(groups);by_order=pd.DataFrame(within)

def safe_rho(x,y):
    x=np.asarray(x,float);y=np.asarray(y,float);ok=np.isfinite(x)&np.isfinite(y);x=x[ok];y=y[ok]
    if len(x)<3 or np.ptp(x)<1e-15 or np.ptp(y)<1e-12:return np.nan
    return float(spearmanr(x,y).statistic)

def fit_exponential(x,y):
    # x is resistant percentage, scaled internally to a fraction.
    x=np.asarray(x,float);y=np.asarray(y,float);ok=np.isfinite(x)&np.isfinite(y)
    x=x[ok];y=y[ok];z=x/100
    if len(x)<4 or np.ptp(x)<1e-10 or np.ptp(y)<1e-10:return None
    def function(z,c,A,K):return c+A*np.exp(-z/K)
    c0=float(np.median(y[z>np.median(z)]));a0=float(np.median(y[z<=np.median(z)])-c0)
    try:
        popt,_=curve_fit(function,z,y,p0=[c0,a0,.1],bounds=([-np.inf,-np.inf,1e-5],[np.inf,np.inf,100]),maxfev=30000)
    except (RuntimeError,ValueError):return None
    pred=function(z,*popt);den=np.sum((y-y.mean())**2)
    return dict(c=popt[0],A=popt[1],K_percent=100*popt[2],r2=1-np.sum((y-pred)**2)/den,
                rmse=float(np.sqrt(np.mean((y-pred)**2))),predict=lambda xx:function(np.asarray(xx)/100,*popt),xmin=x.min(),xmax=x.max())
```

```python
# Composite scores, recovery timing and trade-offs: derived from existing runs.
def score_recovery():
    endpoint=samples[samples.hour.eq(ENDPOINT_HOUR)]
    t=endpoint[endpoint.treated].set_index('scenario')
    c=endpoint[~endpoint.treated].set_index('scenario').reindex(t.index)
    result=t[['order_id','n_resistant','reference','baseline_carrier_share','resistant_share','extinct']].copy()
    result['biomass_component']=[np.clip(ratio(a,b),0,1) for a,b in zip(t.biomass,c.biomass)]
    result['diversity_component']=[np.clip(ratio(a,b),0,1) for a,b in zip(t.shannon,c.shannon)]
    result['composition_component']=1-t.bray_curtis
    w=pd.Series(SCORE_WEIGHTS,dtype=float);w=w/w.sum()
    result['recovery_score']=result[list(w.index)].mul(w,axis=1).sum(axis=1,min_count=3)
    result.loc[result.extinct,'recovery_score']=0.
    result['baseline_resistant_percent']=100*result.baseline_carrier_share
    for penalty in [0.,.25,.5,1.]:
        column=f'score_penalty_{penalty:g}'
        result[column]=result.recovery_score*(1-penalty*result.resistant_share)
        result.loc[result.extinct,column]=0.
    result['adjusted_score']=result.recovery_score*(1-PENALTY_STRENGTH*result.resistant_share)
    result.loc[result.extinct,'adjusted_score']=0.
    return result.reset_index()

def fit_score_candidate(name,data):
    x=data.baseline_carrier_share.to_numpy(float);y=data.recovery_score.to_numpy(float)
    if name=='Linear':
        parameters=np.polyfit(x,y,1)
        return lambda z:np.polyval(parameters,z),parameters
    def saturating(x,c,a,k):return c+(1-c)*a*x/(k+x)
    parameters,_=curve_fit(saturating,x,y,p0=[.5,.8,.05],bounds=([0,0,1e-6],[1,1,100]),maxfev=30000)
    return lambda z:saturating(z,*parameters),parameters

def validate_score_fits(data):
    data=data[~data.reference].dropna(subset=['baseline_carrier_share','recovery_score'])
    rows=[]
    for name in ['Linear','Saturating']:
        errors=[];completed=0
        for order in sorted(data.order_id.unique()):
            train=data[data.order_id.ne(order)];test=data[data.order_id.eq(order)]
            try:
                predict,_=fit_score_candidate(name,train)
                errors.extend(test.recovery_score.to_numpy()-predict(test.baseline_carrier_share.to_numpy()));completed+=1
            except (ValueError,RuntimeError):pass
        rows.append(dict(model=name,held_out_order_RMSE=float(np.sqrt(np.mean(np.square(errors))))
                         if completed==data.order_id.nunique() else np.nan,completed_orders=completed))
    return pd.DataFrame(rows)

def recovery_phase(sim):
    hit=np.flatnonzero(np.isclose(sim['t'],TRANSFER_HOUR,atol=1e-7,rtol=0))
    if not len(hit):raise ValueError('The transfer boundary is missing.')
    start=hit[-1]  # Last record at 24 h is POST-dilution.
    time,indices=np.unique(sim['t'][start:],return_index=True)
    return time,sim['N'][start:][indices]

def timing_profile(x):
    B=x.sum(axis=1);p=np.divide(x,B[:,None],out=np.zeros_like(x),where=B[:,None]>0)
    logp=np.zeros_like(p);np.log(p,out=logp,where=p>0)
    H=-(p*logp).sum(axis=1);H[B<=0]=np.nan
    return B,H,p

def recovery_timing(biomass_threshold=BIOMASS_THRESHOLD,shannon_threshold=SHANNON_THRESHOLD,bc_threshold=BC_THRESHOLD):
    rows=[]
    for d in design.itertuples():
        ts=RUNS[(d.scenario,True)];cs=RUNS[(d.scenario,False)]
        tt,xt=recovery_phase(ts);tc,xc=recovery_phase(cs);end=min(tt[-1],tc[-1])
        grid=TRANSFER_HOUR+np.arange(int(np.floor((end-TRANSFER_HOUR)/CHECK_DT+1e-7))+1)*CHECK_DT
        grid=np.minimum(grid,end)
        a=np.column_stack([np.interp(grid,tt,xt[:,i]) for i in range(57)])
        b=np.column_stack([np.interp(grid,tc,xc[:,i]) for i in range(57)])
        B,H,p=timing_profile(a);Bc,Hc,_=timing_profile(b)
        base=snapshot(ts,BASELINE_HOUR);baseB=base.sum()
        BC=.5*np.abs(p-base/baseB).sum(axis=1);BC[B<=0]=np.nan
        good=(B>0)&(Bc>0)&(Hc>0)&np.isfinite(H)&np.isfinite(BC)
        good&=(B>=biomass_threshold*Bc)&(H>=shannon_threshold*Hc)&(BC<=bc_threshold)&(B>=baseB)
        start=confirmation=np.nan
        for i in np.flatnonzero(good):
            j=np.searchsorted(grid,grid[i]+HOLD_HOURS-1e-8)
            if j<len(grid) and grid[j]-grid[i]>=HOLD_HOURS-1e-8 and good[i:j+1].all():
                start=grid[i]-TRANSFER_HOUR;confirmation=grid[j]-TRANSFER_HOUR;break
        if B[-1]<=0:score=0.;resistant=np.nan
        else:
            components=np.array([np.clip(ratio(B[-1],Bc[-1]),0,1),np.clip(ratio(H[-1],Hc[-1]),0,1),1-BC[-1]])
            score=float(np.average(components,weights=list(SCORE_WEIGHTS.values())))
            resistant=float(a[-1,ts['carrier']].sum()/B[-1])
        rows.append(dict(scenario=d.scenario,order_id=d.order_id,n_resistant=d.n_resistant,reference=d.reference,
            recovered=np.isfinite(start),recovery_time_h=start,confirmation_time_h=confirmation,
            observed_followup_h=grid[-1]-TRANSFER_HOUR,actual_endpoint_hour=grid[-1],endpoint_recovery_score=score,
            endpoint_resistant_fraction=resistant,endpoint_BC=BC[-1],endpoint_biomass=B[-1],meets_thresholds_at_end=bool(good[-1])))
    return pd.DataFrame(rows)

def pareto_choices(times):
    eligible=times[times.recovered&times.meets_thresholds_at_end].dropna(subset=['recovery_time_h','endpoint_recovery_score','endpoint_resistant_fraction']).copy()
    eligible['pareto_optimal']=False
    if len(eligible):
        t=eligible.recovery_time_h.to_numpy();q=eligible.endpoint_recovery_score.to_numpy();r=eligible.endpoint_resistant_fraction.to_numpy();eps=1e-10
        eligible['pareto_optimal']=[not np.any(((t<=t[i]+eps)&(q>=q[i]-eps)&(r<=r[i]+eps))&
            ((t<t[i]-eps)|(q>q[i]+eps)|(r<r[i]-eps))) for i in range(len(eligible))]
    preferred=eligible[(eligible.endpoint_recovery_score>=MIN_ENDPOINT_SCORE)&(eligible.endpoint_resistant_fraction<=MAX_ENDPOINT_RESISTANCE)].sort_values(
        ['recovery_time_h','endpoint_resistant_fraction','endpoint_recovery_score'],ascending=[True,True,False])
    return eligible,preferred

def threshold_sensitivity():
    rows=[]
    for label,b,h,bc in [('Looser',.8,.85,.3),('Default',.9,.9,.2),('Stricter',.95,.95,.1)]:
        z=recovery_timing(b,h,bc)
        rows.append(dict(definition=label,biomass_threshold=b,shannon_threshold=h,BC_threshold=bc,
            configurations=len(z),confirmed_recovery=int(z.recovered.sum()),
            still_meets_at_endpoint=int((z.recovered&z.meets_thresholds_at_end).sum()),
            median_time_among_recovered_h=z.recovery_time_h.median()))
    return pd.DataFrame(rows)

def plot_composite_scores():
    data=recovery_scores[~recovery_scores.reference];counts=INTERMEDIATE_COUNTS
    fig,axes=plt.subplots(1,2,figsize=(14,6),layout='constrained')
    for ax,metric,title in zip(axes,['recovery_score','adjusted_score'],['Recovery score',f'Resistance-adjusted score (penalty strength {PENALTY_STRENGTH:g})']):
        m=data.pivot(index='order_id',columns='n_resistant',values=metric).reindex(index=ORDER_IDS,columns=counts).to_numpy()
        cm=plt.get_cmap('viridis').copy();cm.set_bad('#dddddd')
        im=ax.imshow(np.ma.masked_invalid(m),aspect='auto',cmap=cm,vmin=0,vmax=1)
        ax.set_xticks(range(len(counts)),counts);ax.set_yticks(range(N_ORDERS),[f'Order {i}' for i in ORDER_IDS]);ax.set(title=title,xlabel='Resistant strain count',ylabel='Assignment order')
        for i in range(len(m)):
            for j in range(len(counts)):
                ax.text(j,i,f'{m[i,j]:.2f}' if np.isfinite(m[i,j]) else 'Undefined',ha='center',va='center',fontsize=8,bbox=dict(facecolor='white',alpha=.75,edgecolor='none',pad=.7))
    fig.colorbar(im,ax=axes.tolist(),shrink=.8,label='Score (0 to 1)');fig.suptitle('Recovery quality versus preference for less endpoint resistance',fontsize=15)
    show(fig,'15_composite_scores')
    note(f'Across intermediate assignments, raw recovery scores range from {data.recovery_score.min():.3f} to {data.recovery_score.max():.3f}; resistance-adjusted scores range from {data.adjusted_score.min():.3f} to {data.adjusted_score.max():.3f}. These answer different questions: biological preservation versus a chosen low-resistance objective.')
    display(recovery_scores[recovery_scores.reference].round(4))

def plot_score_scaling():
    fig,axes=plt.subplots(1,2,figsize=(15,5.5),sharey=True,layout='constrained')
    data=recovery_scores[~recovery_scores.reference]
    for ax,metric,title in zip(axes,['recovery_score','adjusted_score'],['Unpenalized recovery','Low-resistance preference score']):
        for order,g in data.groupby('order_id'):ax.scatter(g.baseline_resistant_percent,g[metric],color=ORDER_COLORS[order],s=38,label=f'Order {order}',alpha=.85)
        refs=recovery_scores[recovery_scores.reference];ax.scatter(refs.baseline_resistant_percent,refs[metric],color='#555555',marker='s',s=45,label='Shared reference')
        ax.set(title=title,xlabel='Pretreatment resistant relative abundance (%)',ylabel='Score',xlim=(-2,102),ylim=(-.03,1.03));ax.grid(alpha=.15)
    valid=score_validation.dropna(subset=['held_out_order_RMSE'])
    if len(valid):
        best=valid.sort_values('held_out_order_RMSE').iloc[0]
        fitdata=data.dropna(subset=['baseline_carrier_share','recovery_score']);predict,pars=fit_score_candidate(best.model,fitdata)
        xx=np.linspace(fitdata.baseline_carrier_share.min(),fitdata.baseline_carrier_share.max(),300);axes[0].plot(100*xx,predict(xx),'k--',lw=2,label=best.model+' fit')
        if best.model=='Linear':expression=f'R = {pars[1]:.3f} {pars[0]/100:+.5f}x'
        else:expression=f'R = {pars[0]:.3f} + {(1-pars[0])*pars[1]:.3f}x / ({100*pars[2]:.3f} + x)'
        axes[0].text(.03,.97,expression+f'\nHeld-out-order RMSE = {best.held_out_order_RMSE:.4f}',transform=axes[0].transAxes,va='top',fontsize=9,bbox=dict(facecolor='white',alpha=.85,edgecolor='none'))
        note(f'{best.model} has the lower held-out-order error ({best.held_out_order_RMSE:.4f} score units). This validates prediction across sampled carrier identities for this count grid, not across new ecological systems. No function is fitted to the penalized score as an independent biological relationship.')
    axes[1].text(.03,.04,'Resistance is part of this score by definition.\nIts slope is not an independent biological finding.',transform=axes[1].transAxes,fontsize=9,bbox=dict(facecolor='white',alpha=.9,edgecolor='none'))
    handles,labels=axes[0].get_legend_handles_labels();fig.legend(handles,labels,loc='outside lower center',ncol=6,frameon=False,fontsize=8)
    fig.suptitle('Does resistance predict recovery quality?',fontsize=15);show(fig,'16_recovery_score_scaling')
    display(score_validation.round(5));display(penalty_sensitivity.round(4))

def plot_recovery_timing():
    d=recovery_times[~recovery_times.reference];m=d.pivot(index='order_id',columns='n_resistant',values='recovery_time_h').reindex(index=ORDER_IDS,columns=INTERMEDIATE_COUNTS).to_numpy()
    finite=m[np.isfinite(m)];lo=float(finite.min()) if len(finite) else 0.;hi=float(finite.max()) if len(finite) else 1.
    if np.isclose(lo,hi):lo=max(0,lo-.05);hi+=.05
    fig,ax=plt.subplots(figsize=(11,6),layout='constrained');cm=plt.get_cmap('viridis_r').copy();cm.set_bad('#dddddd')
    im=ax.imshow(np.ma.masked_invalid(m),aspect='auto',cmap=cm,vmin=lo,vmax=hi)
    ax.set_xticks(range(len(INTERMEDIATE_COUNTS)),INTERMEDIATE_COUNTS);ax.set_yticks(range(N_ORDERS),[f'Order {i}' for i in ORDER_IDS])
    ax.set(title='Time to a sustained recovery interval after the 24 h transfer\nNC = not confirmed; numeric differences have 0.1 h evaluation resolution',xlabel='Resistant strain count',ylabel='Assignment order')
    for i in range(len(m)):
        for j in range(len(INTERMEDIATE_COUNTS)):
            ax.text(j,i,f'{m[i,j]:.1f} h' if np.isfinite(m[i,j]) else 'NC',ha='center',va='center',fontsize=9,bbox=dict(facecolor='white',alpha=.75,edgecolor='none'))
    fig.colorbar(im,ax=ax,label='Hours after transfer (lower is faster)');show(fig,'17_time_to_recovery')
    confirmed=recovery_times[recovery_times.recovered]
    note(f'{len(confirmed)}/{len(recovery_times)} design configurations confirm recovery. Among those, recovery starts {confirmed.recovery_time_h.min():.1f}–{confirmed.recovery_time_h.max():.1f} h after transfer. Non-confirmed cases are not assigned a numeric recovery time. This is return to the specified baseline thresholds, not to maximum biomass.')
    display(recovery_times[recovery_times.reference].round(4))

def plot_recovery_tradeoffs():
    fig,ax=plt.subplots(figsize=(11,6),layout='constrained')
    confirmed=recovery_times[recovery_times.recovered].dropna(subset=['endpoint_resistant_fraction'])
    points=ax.scatter(confirmed.recovery_time_h,confirmed.endpoint_recovery_score,c=100*confirmed.endpoint_resistant_fraction,cmap='viridis_r',vmin=0,vmax=100,s=60,alpha=.85)
    frontier=pareto[pareto.pareto_optimal]
    ax.scatter(frontier.recovery_time_h,frontier.endpoint_recovery_score,facecolors='none',edgecolors='black',s=140,lw=1.5,label='Pareto-optimal; still meets endpoint criteria')
    nc=recovery_times[~recovery_times.recovered]
    ax.scatter(nc.observed_followup_h,nc.endpoint_recovery_score,marker='>',color='gray',s=60,label='Not confirmed; position is observation end')
    ax.set(xlabel='Hours after transfer to recovery (lower is faster)',ylabel='Endpoint recovery score (higher is better)',ylim=(-.03,1.03),title='Recovery speed, quality and resistance burden')
    ax.grid(alpha=.15);ax.legend(fontsize=8);fig.colorbar(points,ax=ax,label='Endpoint resistant biomass (%)');show(fig,'18_recovery_tradeoffs')
    note(f'{len(pareto)} configurations both recover and still meet thresholds at the endpoint; {int(pareto.pareto_optimal.sum())} are Pareto-optimal. A black ring means no tested eligible alternative is at least as good on all three objectives and strictly better on one. It is not a universally optimal microbiome.')
    if len(preferred):
        best=preferred.iloc[0]
        note(f'The fastest configuration meeting the explicit quality/resistance targets is {best.scenario}: {best.recovery_time_h:.1f} h, score {best.endpoint_recovery_score:.3f}, and {best.endpoint_resistant_fraction:.1%} endpoint resistance. Ties are broken by lower resistance, then higher score. Selection is resolution- and threshold-dependent.')
        display(preferred.head(10).round(4))
        print('Selected resistant strains:',[NAMES[i] for i in np.flatnonzero(MASKS[best.scenario])])
    else:note('No configuration meets all selection targets. Targets were not automatically relaxed. The Pareto set still shows the available compromises.')
    display(recovery_sensitivity.round(3))
```

**Read it:** orange cells are resistant; gray cells are susceptible. Columns are strain identities. Rows are different assignment orders, all with the same carrier count. Read down a column to see whether that strain changes status between trials.

**Purpose:** verify that the trials change resistance identities without redrawing strain traits. This is a design visualization, not an outcome or a diversity score.

**Calculation:** total height is $B(t)=\sum _iN_i(t)$. Each colored band has vertical thickness $N_i(t)$; its share of the height is $p_i(t)=N_i(t)/B(t)$. The integrated area of a band is not a relative abundance.

**Read it:** each row is one carrier count; left is untreated and right is treated. All panels use the same scale and strain colors. Black traces total biomass. The small initial inoculum is intentional. The sharp 24 h drop is the imposed dilution, not direct antibiotic killing.

**Calculation:** each curve uses actual total biomass $B$, Shannon $H=-\sum  p_i\ln  p_i$, or Gini–Simpson $S=1-\sum  p_i^2$ at each recorded time. No averaging is done in this figure.

## 13. Composite recovery: quality versus the low-resistance objective

**Calculation.** At 48 h, compare biomass and Shannon with the same configuration's untreated control. Compare composition with that treated configuration's own 4 h pretreatment baseline:

$$b=\min(B_T/B_C,1),\quad h=\min(H_T/H_C,1),\quad c=1-BC(48,4),\quad R=(b+h+c)/3$$

Each component ranges from 0 to 1 with the default model outputs. The default weights are equal. Ratios are capped at 1 so extra biomass or diversity cannot compensate indefinitely for a poor result elsewhere. Extinction is explicitly assigned R = 0; other undefined components remain undefined rather than being silently skipped.

To favor lower resistance remaining, calculate:

$$R_{\mathrm{adjusted}}=R(1-\lambda f_{R,48})$$

f_R,48 is the endpoint resistant fraction on a 0–1 scale. With λ = 1, the adjusted score is raw recovery multiplied by susceptible fraction. An all-resistant community therefore scores zero even if it regrows well. λ = 0 gives no penalty; λ = 0.5 is gentler.

**Read it.** Each square is an order/count configuration. Raw recovery and adjusted recovery use the same 0–1 color scale. References are shown once in the table.

**Meaning.** The first score describes recovery; the second expresses a chosen preference. Neither is a validated universal resilience index. Equal weights, capping and the penalty are analysis choices. Shannon and composition partly overlap, so retain their component plots. A relationship with a penalized score is partly built into its definition.

## 15. Recovery speed: when does a community return to baseline?

**The recovery clock starts immediately after the 24 h transfer.** A community must satisfy all four requirements:

1. B_T(t) ≥ 0.90 B_C(t): at least 90% of matched untreated biomass at the same time.
2. H_T(t) ≥ 0.90 H_C(t): at least 90% of matched untreated Shannon at the same time.
3. BC_T(t,4) ≤ 0.20: composition sufficiently close to its own pretreatment baseline.
4. B_T(t) ≥ B_T(4): actual biomass rebuilt to its own pretreatment level.

The fourth condition prevents two small post-transfer communities from qualifying merely because their ratio is near 1. It asks for pretreatment biomass, not maximum carrying capacity.

Let G(t) mean all requirements hold. Recovery time is the first start u after transfer for which G remains true at every evaluated point from u to u + 2 hours:

$$T_{\mathrm{recovery}}=\min\{u\ge0:G(24+v)=1\ \mathrm{for\ all}\ v\in[u,u+2]\}$$

The start time and the later confirmation time are stored separately. A reported 2.8 h means the interval starts at experiment hour 26.8 and is confirmed at 28.8.

**Implementation.** The last saved record at 24 h is the post-dilution state; the earlier one is not used to start recovery. Each strain's absolute biomass is linearly interpolated onto a 0.1 h grid over the common observed treated/control interval. Relative abundance, Shannon and Bray–Curtis are recalculated after interpolation. The original curves can end at 47.9 h, so the analyzed follow-up can be 23.9 h, not a fabricated 24 h. A smaller grid does not recover information absent from the stored trajectories.

**Read it.** Rows are orders, columns are carrier counts, and values are hours after transfer. Gray NC means no qualifying two-hour interval was confirmed before observation ended. The color scale spans observed recovery times to make small differences visible. Differences of one grid step should not be overinterpreted.

**Meaning.** Never-confirmed configurations are not given a recovery time of zero or the observation endpoint. Recovery can later reverse; endpoint criterion status is recorded separately. Thresholds are chosen definitions, not universal biological cutoffs.

## 16. Find promising quality–speed–resistance compromises

**Three objectives.** Prefer a shorter recovery time, a higher unpenalized endpoint score R, and a lower endpoint resistant fraction. Resistance is not also penalized inside R here, which would count the same preference twice.

**Eligibility.** A configuration must have confirmed a two-hour recovery interval and still satisfy the recovery thresholds at the endpoint. It need not have remained continuously recovered between those two checks. The full timing table preserves both statuses.

**Pareto rule.** Configuration A dominates B if A is no slower, has no lower score and has no higher resistant fraction, with at least one strict improvement. A non-dominated configuration is Pareto-optimal among the tested eligible configurations. Several choices can be valid compromises; a ring is not a unique winner or a universal optimum.

**Read it.** X is time after transfer, y is recovery quality, and color is endpoint resistance. Black rings indicate Pareto choices. Gray right-facing triangles show non-confirmed cases at their observation boundary; their horizontal position is not a measured recovery time.

**Preferred choice.** Among eligible configurations with R ≥ 0.90 and endpoint resistant fraction ≤0.20, select the fastest. Ties are broken by lower resistance, then higher score. The exact carrier identities are printed. If none qualifies, report that; do not automatically loosen the target.

**Threshold sensitivity.** Recompute with looser (biomass 0.80, Shannon 0.85, BC 0.30), default (0.90, 0.90, 0.20), and stricter (0.95, 0.95, 0.10) thresholds. Each retains the pretreatment biomass requirement and two-hour hold. Median times are conditional on recovery; always compare the number recovered too.

**Meaning.** This optimizes resistance assignment within one fixed 57-strain community, not species membership, and not a universally optimal microbiome.
